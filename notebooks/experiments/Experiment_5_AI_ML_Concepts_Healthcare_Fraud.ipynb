### 📚 Theoretical Background
Artificial Intelligence (AI) encompasses systems that perceive, reason, and act to achieve goals.
In Health Insurance Fraud Detection, AI has evolved through distinct generations:

| Generation | Core Technology | Strengths | Limitations |
|---|---|---|---|
| **1st Gen** | Static Rule Engines (Expert Systems) | 100% Transparent, Deterministic | Easily circumvented, brittle, high false positives |
| **2nd Gen** | Classical Supervised ML (Trees, Logistic Reg) | Captures multi-variable interactions | Struggles with relational syndicates, black-box opacity |
| **3rd Gen** | Graph Neural Networks (Bipartite graphs) | Uncovers cross-hospital collusion rings | Computationally heavy, hard to explain in court |
| **4th Gen (Syght)** | Multi-Agent Systems + Generative AI + XAI | Real-time, explainable, autonomous triage, audit briefs | Requires careful prompt guardrails & governance |

### Key Concepts:
- **Bias-Variance Tradeoff**: High bias = underfitting (Rule baselines); High variance = overfitting (unpruned deep trees).
- **Algorithmic Fairness**: Never using protected patient demographics (gender, race, caste, religion) as features. Only provider billing behavior is audited!

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Simulation of the Bias-Variance Tradeoff across model complexity
complexity = np.linspace(1, 10, 50)
bias_squared = 1.0 / (0.5 * complexity + 0.5)
variance = 0.05 * (complexity ** 1.8)
irreducible_error = 0.15
total_error = bias_squared + variance + irreducible_error

plt.figure(figsize=(9, 5))
plt.plot(complexity, bias_squared, label='Bias²', color='#3b82f6', lw=2)
plt.plot(complexity, variance, label='Variance', color='#ef4444', lw=2)
plt.plot(complexity, total_error, label='Total Expected Error', color='#7928ca', lw=2.5, linestyle='--')
plt.axvline(x=complexity[np.argmin(total_error)], color='gray', linestyle=':', label='Sweet Spot (Optimal Model Complexity)')
plt.title('The Bias-Variance Tradeoff in Healthcare Fraud Classifiers')
plt.xlabel('Model Capacity / Complexity (Rules -> Linear -> Ensembles -> Deep Nets)')
plt.ylabel('Error')
plt.legend()
plt.tight_layout()
plt.show()
```

### Ethical AI & Demographic Fairness Verification

```python
from ml_pipeline.config import SyghtConfig
cfg = SyghtConfig()

# Audit engineered feature list for patient demographic exclusion
from ml_pipeline.feature_engineering import build_provider_features
from ml_pipeline.medicare_simulator import generate_simulated_medicare_data

lbl, bene, ip, op = generate_simulated_medicare_data(n_providers=20, seed=42)
feats = build_provider_features(lbl, bene, ip, op, config=cfg)

forbidden_demographics = ['Race', 'Gender', 'State', 'County', 'Religion', 'Caste']
violations = [col for col in feats.columns if any(dem.lower() in col.lower() for dem in forbidden_demographics)]

print(f'Demographic Feature Violations Detected: {len(violations)}')
assert len(violations) == 0, 'Algorithmic Fairness Violation: Demographic features detected in feature matrix!'
print('Algorithmic Fairness Audit PASSED: Zero patient demographic bias in provider scoring.')
```
