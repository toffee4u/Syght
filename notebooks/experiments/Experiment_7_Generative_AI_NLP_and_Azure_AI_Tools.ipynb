### 📚 Theoretical Background
Unstructured clinical text hides the most sophisticated healthcare fraud schemes.

**1. Natural Language Processing (NLP)**:
- Extracts symptoms, procedures, and conditions from physician consultation notes and hospital discharge summaries.
- Compares textual medical realities against submitted ICD-10 billing codes to catch **Upcoding** and **Phantom Billing**.

**2. Azure AI Document Intelligence**:
- Deep learning optical character recognition (OCR) and layout analysis for scanned hospital invoices, lab slips, and discharge certificates.

**3. Azure AI Search**:
- Vector and semantic search across medical ICD ontologies and historical de-empanelment case precedents.

```python
from ml_pipeline.nlp_clinical_analyzer import (
    ClinicalNotesNLPAnalyzer,
    AzureAIDocumentIntelligenceConnector,
    AzureAISearchConnector
)

nlp_analyzer = ClinicalNotesNLPAnalyzer()
doc_intelligence = AzureAIDocumentIntelligenceConnector()
search_connector = AzureAISearchConnector()
print('NLP Analyzer and Azure AI Connectors initialized successfully!')
```

### Step 1: NLP Entity Extraction on Clinical Doctor Notes

```python
clinical_note = '''
Patient presented to the outpatient department complaining of mild throat irritation, runny nose, and low-grade fever for 2 days.
Chest auscultation clear, bilateral breath sounds normal. Abdomen soft, non-tender. Patient was oriented and ambulatory.
Diagnosis: Acute viral pharyngitis. Prescribed oral paracetamol 500mg TDS, warm saline gargles, and rest.
'''

entities = nlp_analyzer.extract_clinical_entities(clinical_note)
print('Extracted Clinical Entities from Doctor Narrative:')
for category, terms in entities.items():
    print(f' - {category.capitalize()}: {terms}')
```

### Step 2: Semantic Concordance & Upcoding Detection

```python
# Fraudulent Case: The hospital billed Sepsis (A41) and Heart Failure (I50) for this mild cough!
fraudulent_billed_codes = ['A41', 'I50']

audit_result = nlp_analyzer.verify_notes_against_codes(clinical_note, fraudulent_billed_codes)
print(f'Semantic Concordance Score: {audit_result.semantic_concordance_score:.2f} / 1.00')
print(f'Upcoding Suspected:          {audit_result.upcoding_suspected}')
print(f'Phantom Billing Suspected:   {audit_result.phantom_billing_suspected}\n')
print('FORENSIC AUDIT FINDINGS:')
for finding in audit_result.audit_findings:
    print(f' [!] {finding}')
```

### Step 3: Azure AI Document Intelligence & Search Integration

```python
# 1. Azure AI Document Intelligence (Health Insurance Invoice OCR)
scanned_bill_sample = '''
METROPOLITAN SURGICAL CARE HOSPITAL
Date of Admission: 2026-09-12 | Date of Discharge: 2026-09-28
Patient: Beneficiary_BNE8819 | Total Billed Amount: INR 485,000.00
Procedures: Unbundled Laparoscopic Cholecystectomy + Emergency Tray
'''
doc_result = doc_intelligence.parse_scanned_bill(scanned_bill_sample)
print('Azure AI Document Intelligence Extraction:')
print(doc_result)

# 2. Azure AI Search (Vector Retrieval of Case Precedents)
precedents = search_connector.search_fraud_precedents('unbundling metabolic panel', top_k=2)
print('\nAzure AI Search Precedents Retrieved:')
for p in precedents:
    print(f' - [{p["case_id"]}] {p["title"]} (Relevance: {p["relevance_score"]:.2f})')
```
